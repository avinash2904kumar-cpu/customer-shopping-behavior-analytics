```python
labels = ['Young','Adult','Middle_aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
#qcut splits data into equal group based on the data given and assign the label we provide
```

```python
#create a column (purchase freq days)
```

```python
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthy' : 30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90,
}
df['purchase_freq_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
